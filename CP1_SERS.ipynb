# Dataset 3 — Power Consumption of Tetouan City (UCI)

Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

Situação:

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda


```python
df3 = pd.read_csv('/content/Tetuan City power consumption.csv')
```

```python
df3.columns
```

```python
#Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3

df3 = df3.rename(columns={
    'Temperature': 'Consumo_Zona_1',
    'Humidity': 'Consumo_Zona_2',
    'Wind Speed': 'Consumo_Zona_3'
})
```

```python
#Determine o consumo máximo registrado em cada uma das três zonas.

df3['Consumo_Zona_1'] = pd.to_numeric(df3['Consumo_Zona_1'], errors='coerce')
df3['Consumo_Zona_2'] = pd.to_numeric(df3['Consumo_Zona_2'], errors='coerce')
df3['Consumo_Zona_3'] = pd.to_numeric(df3['Consumo_Zona_3'], errors='coerce')


maximos = df3[['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']].max()
print(maximos)
#Identifique qual zona apresenta o maior pico de consumo na amostra.
zona_maior_pico = maximos.idxmax()
maior_pico = maximos.max()

print("Zona com maior pico:", zona_maior_pico)
print("Valor do maior pico:", maior_pico)
#Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.
valor_maximo = df3[zona_maior_pico].max()
limiar = 0.70 * valor_maximo
df_acima_limiar = df3[df3[zona_maior_pico] > limiar]

print("Limiar (70%):", limiar)
print(df_acima_limiar)
```

```python
#Conte os registros e calcule o percentual que eles representam.

# Conta os registros acima do limiar
quantidade = len(df_acima_limiar)

# Calcula o percentual em relação ao total da amostra
percentual = (quantidade / len(df3)) * 100

print("Quantidade de registros:", quantidade)
print("Percentual:", percentual, "%")
```

```python
#Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.

# Calcula a temperatura média
temperatura_media = df3['Consumo_Zona_1'].mean()

# Cria o DataFrame com as duas condições
df_acima_limiar_temperatura = df3[
    (df3[zona_maior_pico] > limiar) &
    (df3['Consumo_Zona_1'] > temperatura_media)
]

print("Temperatura média:", temperatura_media)
print(df_acima_limiar_temperatura)

```

```python
#Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros após a inclusão da condição ambiental.

# Quantidade antes da condição de temperatura
qtd_antes = len(df_acima_limiar)

# Quantidade depois de adicionar a condição de temperatura
qtd_depois = len(df_acima_limiar_temperatura)

# Diferença
diferenca = qtd_antes - qtd_depois

# Percentual de redução
percentual_reducao = (diferenca / qtd_antes) * 100

print("Registros antes:", qtd_antes)
print("Registros depois:", qtd_depois)
print("Registros removidos:", diferenca)
print(f"Redução: {percentual_reducao:.2f}%")
```

#Dataset 5 — Wind & Solar Energy Production (Kaggle)

```python
df5 = pd.read_csv('/content/Energy Production Dataset.csv')
```

```python
df5 = df5.pivot_table(
    index=['Date', 'Start_Hour', 'End_Hour', 'Day_of_Year', 'Season', 'Month_Name'],
    columns='Source',
    values='Production',
    fill_value=0
).reset_index()
```

```python
df5.rename(columns={
    'Solar': 'Geracao_Solar',
    'Wind': 'Geracao_Eolica',
                    }, inplace=True)
```

```python
df5.head()
```

```python
max_solar = df5['Geracao_Solar'].max()
```

```python
max_eolica = df5['Geracao_Eolica'].max()
```

```python
limiar = 0.70
limite_solar = limiar * max_solar
limite_eolica = limiar * max_eolica
```

```python
df_alta_solar = df5[df5['Geracao_Solar'] >= limite_solar].copy()
```

```python
df_alta_eolica = df5[df5['Geracao_Eolica'] >= limite_eolica].copy()
```

```python
total_registros = len(df5)
qtd_alta_solar = len(df_alta_solar)
qtd_alta_eolica = len(df_alta_eolica)
pct_alta_solar = (qtd_alta_solar / total_registros) * 100
pct_alta_eolica = (qtd_alta_eolica / total_registros) * 100
```

```python
print(f"Total de registros na base: {total_registros:,}")
print("-" * 45)
print(f"Alta Geração Solar  : {qtd_alta_solar:,} registros ({pct_alta_solar:.2f}%)")
print(f"Alta Geração Eólica : {qtd_alta_eolica:,} registros ({pct_alta_eolica:.2f}%)")
```

```python
if pct_alta_solar > pct_alta_eolica:
    diferenca = pct_alta_solar - pct_alta_eolica
    print(f"A fonte SOLAR opera com maior frequência acima de 70% do seu pico "
          f"({pct_alta_solar:.2f}% vs {pct_alta_eolica:.2f}% eólica, diferença de +{diferenca:.2f} p.p.).")
elif pct_alta_eolica > pct_alta_solar:
    diferenca = pct_alta_eolica - pct_alta_solar
    print(f"A fonte EÓLICA opera com maior frequência acima de 70% do seu pico "
          f"({pct_alta_eolica:.2f}% vs {pct_alta_solar:.2f}% solar, diferença de +{diferenca:.2f} p.p.).")
else:
    print(f"Ambas as fontes apresentaram a mesma frequência de operação acima de 70% "
          f"({pct_alta_solar:.2f}%).")
```

Como as usinas solares e eólicas possuem capacidades instaladas e escalas distintas, comparar sua produção por valores fixos em megawatts distorce a análise em favor da fonte de maior porte.     Este codigo esta condisente com oq foi pedido na atividade

#Dataset 6 — Individual Household Electric Power Consumption (UCI)
Fonte: https://archive.ics.uci.edu/dataset/235/individual+household+electric+power+consumption

Situação:

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é o mesmo utilizado como referência em aula, mas o critério de análise será diferente.

```python
df6 = pd.read_csv("/content/energydata_SAMPLE.csv")
```

```python
df6.columns
```

```python
#Carregue a nova amostra e organize os atributos utilizando nomes mais simples.

df6 = df6.rename(columns={
    'Appliances': 'eletrodomesticos',
    'lights': 'luzes',
    'T1': 'temp1',
    'T2': 'temp2',
    'T3':'temp3',
    'RH_1':'unidade1',
    'RH_2': 'unidade2',
    'RH_3': 'unidade3'

})
```

```python
#Determine o valor máximo de potência ativa.
df6["eletrodomesticos"].max()
```

```python
#Calcule 75% do valor máximo e crie um DataFrame contendo registros acima desse limite.

valor_maximo = df6["eletrodomesticos"].max()

limite = 0.75 * valor_maximo

df_acima_limite = df6[df6["eletrodomesticos"] > limite]

print("Valor máximo:", valor_maximo)
print("Limite de 75%:", limite)
print(df_acima_limite)
```

```python
#Calcule a quantidade e o percentual de registros selecionados.

quantidade = len(df_acima_limite)
percentual = (quantidade / len(df6)) * 100

print("Quantidade de registros:", quantidade)
print("Percentual de registros:", round(percentual, 2), "%")
```

```python
#Calcule a corrente média da amostra.

df6["corrente"] = df6["eletrodomesticos"] / 220
corrente_media = df6["corrente"].mean()
```

```python
#Crie um segundo DataFrame contendo simultaneamente potência ativa acima de 75% do máximo e corrente acima da média.
df_filtrado = df6[
    (df6["eletrodomesticos"] > limite) &
    (df6["corrente"] > corrente_media)
]

print(df_filtrado)
```
