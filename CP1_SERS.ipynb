# Dataset 3 — Power Consumption of Tetouan City (UCI)

Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

Situação:

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda


```python
df3 = pd.read_csv('/content/Tetuan City power consumption.csv')
```

```python
df3.columns
```

```python
#Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3

df3 = df3.rename(columns={
    'Temperature': 'Consumo_Zona_1',
    'Humidity': 'Consumo_Zona_2',
    'Wind Speed': 'Consumo_Zona_3'
})
```

```python
#Determine o consumo máximo registrado em cada uma das três zonas.

df3['Consumo_Zona_1'] = pd.to_numeric(df3['Consumo_Zona_1'], errors='coerce')
df3['Consumo_Zona_2'] = pd.to_numeric(df3['Consumo_Zona_2'], errors='coerce')
df3['Consumo_Zona_3'] = pd.to_numeric(df3['Consumo_Zona_3'], errors='coerce')


maximos = df3[['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']].max()
print(maximos)
#Identifique qual zona apresenta o maior pico de consumo na amostra.
zona_maior_pico = maximos.idxmax()
maior_pico = maximos.max()

print("Zona com maior pico:", zona_maior_pico)
print("Valor do maior pico:", maior_pico)
#Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.
valor_maximo = df3[zona_maior_pico].max()
limiar = 0.70 * valor_maximo
df_acima_limiar = df3[df3[zona_maior_pico] > limiar]

print("Limiar (70%):", limiar)
print(df_acima_limiar)
```

```python
#Conte os registros e calcule o percentual que eles representam.

# Conta os registros acima do limiar
quantidade = len(df_acima_limiar)

# Calcula o percentual em relação ao total da amostra
percentual = (quantidade / len(df3)) * 100

print("Quantidade de registros:", quantidade)
print("Percentual:", percentual, "%")
```

```python
#Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.

# Calcula a temperatura média
temperatura_media = df3['Consumo_Zona_1'].mean()

# Cria o DataFrame com as duas condições
df_acima_limiar_temperatura = df3[
    (df3[zona_maior_pico] > limiar) &
    (df3['Consumo_Zona_1'] > temperatura_media)
]

print("Temperatura média:", temperatura_media)
print(df_acima_limiar_temperatura)

```

```python
#Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros após a inclusão da condição ambiental.

# Quantidade antes da condição de temperatura
qtd_antes = len(df_acima_limiar)

# Quantidade depois de adicionar a condição de temperatura
qtd_depois = len(df_acima_limiar_temperatura)

# Diferença
diferenca = qtd_antes - qtd_depois

# Percentual de redução
percentual_reducao = (diferenca / qtd_antes) * 100

print("Registros antes:", qtd_antes)
print("Registros depois:", qtd_depois)
print("Registros removidos:", diferenca)
print(f"Redução: {percentual_reducao:.2f}%")
```
